```python
# hier aus den Stationen, die in der Spalte "@relations" ein json-Dict stehen haben, die Linien-Nummern und Farben extrahieren.
# Danach wieder an den Original-gdf_stations dranjoinen.

# Beispiel:
# für eine Zeile aus dem gdf_stations die verschiedenen Tags + Values schön ausgeben:
# from pprint import pprint
# xyz = gdf_stations[gdf_stations["name"]=='Kröpcke']["@relations"].iloc[0]
# xyz
# pprint(json.loads(xyz))

# Beispiel:
# alle Keys für eine exemplarische Zeile printen (ohne values)
# xyz = gdf_stations[gdf_stations["name"]=='Kröpcke']["@relations"].iloc[0
# def get_all_keys(obj, prefix=""):
#     keys = set()
#     if isinstance(obj, dict):
#         for k, v in obj.items():
#             new_prefix = f"{prefix}.{k}" if prefix else k
#             keys.add(new_prefix)
#             keys |= get_all_keys(v, new_prefix)
#     elif isinstance(obj, list):
#         for i, item in enumerate(obj):
#             keys |= get_all_keys(item, prefix)
#     return keys
# data = json.loads(xyz)
# keys = get_all_keys(data)
# for k in sorted(keys):
#     print(k)


def extract_colour_ref_list(stations):

    gdf_stations = stations[["name","@relations"]].dropna()

    def loads_fuzzy(s):
        if isinstance(s, (list, dict)):
            return s
        try:
            return json.loads(s)
        except Exception:
            return ast.literal_eval(s)


    col = "@relations"
    parsed = gdf_stations[col].apply(loads_fuzzy)
    parsed

    expl = gdf_stations.drop(columns=[col]).join(
        pd.Series(parsed, index=gdf_stations.index, name="parsed")
    ).explode("parsed", ignore_index=False)


    # Felder ziehen
    expl["reltags_colour"] = expl["parsed"].apply(
        lambda d: ((d or {}).get("reltags", {}) or {}).get("colour")
                  or ((d or {}).get("reltags", {}) or {}).get("color")
    )
    expl["reltags_ref"] = expl["parsed"].apply(
        lambda d: ((d or {}).get("reltags", {}) or {}).get("ref")
    )

    # expl[["name","reltags_ref","reltags_colour"]]
    result_long = expl.drop(columns=["parsed"]).drop_duplicates().sort_values("reltags_ref")#
    result_long
    agg_back = (result_long
                .groupby("name", as_index=False)
                .agg(reltags_colours=("reltags_colour", lambda s: [x for x in s if pd.notna(x)]),
                     reltags_refs=("reltags_ref",     lambda s: [x for x in s if pd.notna(x)]))
               )

    agg_back
    df_joined = stations.merge(agg_back, on="name", how="left")
    return df_joined
```

```python
# Stations-Punkte extrahieren
doc = fitz.open("UESTRA_Netzplan-U_2024-25.pdf")
page = doc[0]  # es gibt nur eine Seite
page_rect = page.rect
cdrawings = page.get_cdrawings()#, sort=True)


circles = []
rects = []

# Parameter nach Bedarf anpassen
square_tol = 0.35      # max. 15% Abweichung von Quadrat
min_diam   = 1.85     # minimaler Durchmesser
max_diam   = 20.0      # maximaler Durchmesser

for d in cdrawings:
    # Nur gefüllte Pfade
    # if d.get("type") != "f":
    #     continue
    if d.get("type") not in ("f", "s", "fs"):
        continue

    items = d.get("items", [])
    if not items:
        continue

    #rects.append(d["rect"])
    # Nur Objekte betrachten, deren Items ausschließlich 'c' (Bezierkurven) sind
    if not all(item[0] == "c" for item in items):
        continue

    x0, y0, x1, y1 = d["rect"]
    w = x1 - x0
    h = y1 - y0
    if w <= 0 or h <= 0:
        continue

    # annähernd quadratisch?
    if abs(w - h) > max(w, h) * square_tol:
        continue

    # Größen-Filter (damit du nicht riesige Kreise / Flächen mit erwischst)
    if not (min_diam <= max(w, h) <= max_diam):
        continue

    # Mittelpunkt / Radius aus dem bounding rect
    cx = (x0 + x1) / 2.0
    cy = (y0 + y1) / 2.0
    r  = max(w, h) / 2.0

    if (min_diam <= r <= max_diam):

        #print(f"{cx}, {cy}")
        circles.append((cx,cy))
        rects.append(d["rect"])
        #print(r)
```

## Zusammenbringen von Netzplan und Dataframe

```python
# für jede Zeile aus dem df: schauen, welcher Circle am nähesten ist.
import numpy as np

def dist_point_to_rect_edge(px, py, x0, y0, x1, y1):
    # sicherstellen, dass x0<x1, y0<y1
    if x0 > x1:
        x0, x1 = x1, x0
    if y0 > y1:
        y0, y1 = y1, y0

    inside_x = x0 <= px <= x1
    inside_y = y0 <= py <= y1

    # Punkt liegt im Rechteck → Abstand zum Rand
    if inside_x and inside_y:
        return min(px - x0, x1 - px, py - y0, y1 - py)

    # senkrecht zu einer horizontalen Kante
    if inside_x:
        return y0 - py if py < y0 else py - y1

    # senkrecht zu einer vertikalen Kante
    if inside_y:
        return x0 - px if px < x0 else px - x1

    # Ecke (außerhalb in x und y)
    dx = x0 - px if px < x0 else px - x1
    dy = y0 - py if py < y0 else py - y1
    return (dx**2 + dy**2) ** 0.5

points = circles #[(px1, py1), (px2, py2), ...]  # deine Punktliste

def nearest_point_to_rect_edge(rect_list):

    # falls Station nicht auf der Karte gefunden wurde (dann ist Spalte quad_rect_text=0
    if rect_list==0:
        return None, np.nan

    x0 = rect_list.x0
    y0 = rect_list.y0
    x1 = rect_list.x1
    y1 = rect_list.y1
    dists = [
        dist_point_to_rect_edge(px, py, x0, y0, x1, y1)
        for (px, py) in points
    ]
    idx = int(np.argmin(dists))
    return points[idx], dists[idx]


# rect_coords ist eine Liste [x0, y0, x1, y1] in jeder Zeile
gdf_stations["nearest_point"], gdf_stations["nearest_dist"] = zip(
    *gdf_stations["quad_rect_text"].apply(nearest_point_to_rect_edge)
)
```

```python
# Maske: Zeilen, in denen GPS vorhanden ist (Basis für die Kalibrierung)
mask_known = gdf_stations["nearest_point"].notna()

# Individuelle Koordinaten zerlegen

# df['B'].apply(lambda p: [y[0] for y in p])

x_indiv = gdf_stations.loc[mask_known, "nearest_point"].str[0]#apply(lambda p: [y[0] for y in p])#.apply(lambda p: p[0])
y_indiv = gdf_stations.loc[mask_known, "nearest_point"].str[1]#apply(lambda p: [y[0] for y in p])#.apply(lambda p: p[0])
# y_indiv = gdf_stations.loc[mask_known, "nearest_point"].apply(lambda p: p[1])

# GPS zerlegen
lon = gdf_stations.loc[mask_known, "center"].x#apply(lambda p: p[0])
lat = gdf_stations.loc[mask_known, "center"].y# apply(lambda p: p[1])

# Min / Max der individuellen Skala
x_min, x_max = x_indiv.min(), x_indiv.max()
y_min, y_max = y_indiv.min(), y_indiv.max()

# Min / Max der GPS-Skala
lon_min, lon_max = lon.min(), lon.max()
lat_min, lat_max = lat.min(), lat.max()


def map_x_to_lon(x):
    if x_max == x_min:
        return np.nan   # degenerierter Fall, zur Sicherheit
    return lon_min + (x - x_min) * (lon_max - lon_min) / (x_max - x_min)

def map_y_to_lat(y):
    if y_max == y_min:
        return np.nan
    return lat_min + (y - y_min) * (lat_max - lat_min) / (y_max - y_min)


def map_indiv_to_gps(indiv_pt):
    if not isinstance(indiv_pt, (list, tuple)) or len(indiv_pt) != 2:
        return np.nan#, np.nan)
    x, y = indiv_pt

    # p = Point(1.0, -1.0)

    # geometry = geopandas.points_from_xy(df.longitude, df.latitude, crs="EPSG:4326")
    return shapely.Point(map_x_to_lon(x), map_y_to_lat(y))

gdf_stations["netzplan_gps_mapped"] = gdf_stations["nearest_point"].apply(map_indiv_to_gps)

```
